```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python
# @title Task 1.1 - Generate elementary signals
n = np.arange(-5, 11)
step = (n >= 0).astype(float)

values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, values):
    x = (a**n) * step
    ax.stem(n, x, basefmt="k-")
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("x(n)")

plt.tight_layout()
plt.show()
```

```python
# @title Task 1.2 - (a) Plot exponential signals
n = np.arange(-5, 11)
step = (n >= 0).astype(float)

values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, values):
    x = (a**n) * step
    ax.stem(n, x, basefmt="k-")
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("x(n)")

plt.tight_layout()
plt.show()
```

```python
# @title Task 1.2 - (b) Plot cosine signals
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))

plt.stem(n, x1, linefmt="b-", markerfmt="bo",
         basefmt=" ", label="cos(2πn/8)")

plt.stem(n, x2, linefmt="r-", markerfmt="rx",
         basefmt=" ", label="cos(2πn/7.5)")

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Discrete-Time Cosine Signals")
plt.legend()
plt.grid()
plt.show()
```

```python
# @title Task 1.3 — Working with audio
import os

fname = "my_voice.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    if x.ndim > 1:
        x = x.mean(axis=1)

    x = x / np.max(np.abs(x))

else:
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs

print(f"fs = {fs} Hz")
print(f"duration = {len(x)/fs:.2f} s")
print(f"samples = {len(x)}")

plt.plot(t, x)
plt.xlabel("time (s)")
plt.ylabel("amplitude")
plt.title("My recording")
plt.show()
```

2.  Why is a stem plot more appropriate than a continuous line for 𝑥(𝑛)?

A stem plot is more appropriate because a discrete-time signal is defined only at integer values of \(n\). A stem plot shows the individual samples clearly without implying that the signal exists continuously between sample points.
