ask 3.1 — Convolution from the definition

```python
import numpy as np
import time

def conv(x, h):
    N = len(x) + len(h) - 1
    y = np.zeros(N)

    for n in range(N):
        for k in range(len(x)):
            if 0 <= n - k < len(h):
                y[n] += x[k] * h[n - k]

    return y
```

```python
Task 3.3 — Testing linearity and time-invariance numerically
```

```python
rng = np.random.default_rng(0)

x1 = rng.standard_normal(50)
x2 = rng.standard_normal(50)

a, b = 2.0, -3.0

systems = {
    "x^2": lambda x: x**2,
    "n x(n)": lambda x: np.arange(len(x)) * x,
    "moving average": lambda x: np.convolve(x, np.ones(3)/3)[:len(x)]
}

for name, T in systems.items():
    linear = np.allclose(
        T(a*x1 + b*x2),
        a*T(x1) + b*T(x2)
    )

    xd = np.concatenate([np.zeros(5), x1])[:50]
    yd = np.concatenate([np.zeros(5), T(x1)])[:50]

    tinv = np.allclose(T(xd), yd)

    print(f"{name:15s} linear={linear} time-invariant={tinv}")
```

1. **x²:** The system is non-linear because squaring the input does not satisfy the superposition property. It is time-invariant because delaying the input produces the same delay in the output.

2. **n x(n):** The system is linear because it satisfies the superposition property. However, it is time-varying because its output depends explicitly on the time index n.

3. **Moving average:** The system is linear because it satisfies the superposition property. It is also time-invariant because delaying the input produces an equally delayed output.

Therefore, the moving-average system is the only LTI system among the three tested systems.
